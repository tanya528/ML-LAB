```python
#Q1.(b)Create a 2d array use seed command, slice array into A,B,C
# ,find sum of A C and product of A and B.

import numpy as np
np.random.seed(45)   #set seed for same output

#create 10x10 array
arr=np.random.randint(1,100,size=(10,10))

#Slice array A(5x7)
A=arr[0:5,0:7]
print(" Array A:" ,A,"\n")

#Slice array B(7X5)
B=arr[0:7,0:5]
print("Array B:",B,"\n")

#Slice array C(5x7)
C=arr[0:5,3:10]
print("Array C:",C,"\n")

#sum of A & C
sum_AC=A+C
print("Sum of A & C:" ,sum_AC,"\n")

#product of A & B
mul_AB=np.dot(A,B)
print("Multiplication of A & B:" ,mul_AB,"\n")
```

```python
#Q2.Create a 10x10 2d array and find all possible reshaped 2 and 3d arrays from it


import numpy as np

np.random.seed(30)

arr2 = np.random.randint(1, 100, size=(10, 10))

print("Original array:")
print(arr2)

# All possible 2D arrays
print("\n2D arrays:")

for i in range(1, 101):
    if 100 % i == 0:
        print("Shape:", (i, 100 // i))
        print(arr2.reshape(i, 100 // i))

# All possible 3D arrays
print("\n3D arrays:")

for i in range(1, 101):
    for j in range(1, 101):
        if 100 % (i * j) == 0:
            k=100 // (i * j)
            print("Shape:", (i, j, k))
            print(arr2.reshape(i, j, k))
```
